## 07 Sum of weights

```python
asia_set = df[(df['origin'] == 'Asia')][['vehicle_weight','model_year']]
asia_set
```

```python
asia_weight_year = asia_set[['vehicle_weight', 'model_year']]
X = asia_set.head(7).to_numpy()
X

```

```python
X_transposed = X.T
X_transposed
```

```python
X_transposed.shape
```

```python
XTX = np.dot(X, X_transposed)
XTX
```

```python
XTX.shape
```

```python
XTX_invert = np.linalg.pinv(XTX)
XTX_invert
```
